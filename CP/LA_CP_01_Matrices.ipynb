Знайдіть відповідні функції в numpy. Подивіться їх вихідний код та порівняйте з вашим.

```python
#write your code here
import pandas as pd
import numpy as np

def is_symmetric_matrix(arr):
    arr_temp = arr.transpose()
    return np.allclose(arr, arr_temp)

def is_upper_triangular_matrix(arr):
    result = True

    for i_r, row in enumerate(arr):
      for i_c, col in enumerate(row):
        if i_r > 0 and i_c < i_r:
          if col != 0:
            result = False

    return result

def is_lower_triangular_matrix(arr):
    result = True

    for i_r, row in enumerate(arr):
      for i_c, col in enumerate(row):
        if i_r < len(arr) and i_c > i_r:
          if col != 0:
            result = False

    return result

def is_diagonal_matrix(arr):
    result = True

    for i_r, row in enumerate(arr):
      for i_c, col in enumerate(row):
        if i_r != i_c:
          if col != 0:
            result = False

    return result


def is_scalar_matrix(arr):
    n = len(arr)
    result = True
    for i in range(n):
        for j in range(n):
            if i == j:
                if arr[i][j] != arr[0][0]:
                    result = False
            else:
                if arr[i][j] != 0:
                    result = False

    return result

arr = np.array([
        [1, 0, 1],
        [0, 2, 5],
        [1, 5, 3]
        ])

arr2 = np.array([
        [1, 0, 1],
        [0, 2, 5],
        [0, 0, 3]
        ])

arr3 = np.array([
        [1, 0, 0],
        [0, 2, 0],
        [0, 0, 3]
        ])

arr4 = np.array([
        [2, 0, 0],
        [0, 2, 0],
        [0, 0, 2]
        ])

print(is_symmetric_matrix(arr))
print(is_upper_triangular_matrix(arr2))
print(is_lower_triangular_matrix(arr2.transpose()))
print(is_diagonal_matrix(arr3))
print(is_scalar_matrix(arr4))

```

💻 **1.2. Діагоналі.**

Напишіть функцію diag\_matrix, яка отримує на вхід натуральне число $n$ і будує матрицю, на діагоналях якої стоять відповідно $0,1,2, \dots n, n-1, \dots 0$ . Наприклад, для $ n = 3 $ будує матрицю
$\left( \begin{array}{cccc} 3 & 2 & 1 & 0 \\ 2 & 3 & 2 & 1 \\ 1 & 2 & 3 & 2 \\ 0 & 1 & 2 & 3 \end{array} \right)$. Використовуйте засоби numpy, не використовуйте цикли та поелементне присвоєння.

```python
#write your code here
import numpy as np

def diag_matrix(n):
    indices = np.arange(n + 1)

    diff_matrix = np.abs(indices[:, np.newaxis] - indices)

    result = n - diff_matrix

    return result

print(diag_matrix(3))
```

1. $(A^t)^t=A$;
2. $(A+B)^t=A^t+B^t$;
3. $(A-B)^t=A^t-B^t$;
4. $(\lambda A)^t=\lambda A^t$;
5. $(AB)^t=B^tA^t$;
6. $(AB)^{-1}=B^{-1}A^{-1}$;
7. $(A^t)^{-1}=(A^{-1})^t$.


```python
#write your code here
import numpy as np

A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
l = 5
# 1. Обмежень немає.
print(np.allclose((A.T).T, A))

# 2. Матриці повинні мати однаковий розмір
print(np.allclose((A + B).T, A.T + B.T))

# 3. Матриці повинні мати однаковий розмір
print(np.allclose((A - B).T, A.T - B.T))

# 4. Обмежень немає.
print(np.allclose((l * A).T, l * A.T))

# 5.
print(np.allclose((A @ B).T, B.T @ A.T))

# 6.
A_inv = np.linalg.inv(A)
B_inv = np.linalg.inv(B)
print(np.allclose(np.linalg.inv(A @ B), B_inv @ A_inv))

# 7.
print(np.allclose(np.linalg.inv(A.T), (np.linalg.inv(A)).T))
```

💻 **1.7. Маркетингова кампанія.** Компанія влаштовує рекламну кампанію своєї продукції. Для цього усіх потенційних клієнтів поділено на $m$ груп (кожна група характеризується спільними характеристиками, наприклад, чоловіки 35-45 років з вищою освітою або жінки-домогосподині віком понад 60, тощо). Компанія замовляє рекламу продукції за $n$ каналами (наприклад, на телебаченні, на веб-сайтах, у відео-блогах, поштою тощо). Здатність каналів забезпечити перегляди потенційних клієнтів характризується $m\times n$ *матрицей охоплення*  $R=(R_{ij})$, де $R_{ij}$ -- це кількість переглядів клієнтів із групи $i$ на кожен долар (або іншу грошову одиницю), витрачений на рекламу в каналі $j$. Наприклад, якщо на рекламу на ТБ-шоу (канал 1) було витрачено 1000 доларів і це шоу подивилось 30 000 домогосподарок пенсійного віку (група 1) і 5 студентів-чоловіків віком до 25 років (група 2), то $R_{11}=30$, $R_{21} = 0.005$.

Нехай вектор $c$ з $n$ компонентами характеризує витрати компанії на рекламу за $n$ каналами, а вектор $v$ з $m$ компонентами характеризує охоплення кожної з $m$ груп клієнтів за всіма каналами реклами, тобто $v_i$ -- це загальна кількість переглядів реклами за всіма каналами клієнтами з групи $i$. Також розглянемо вектор  $a$ з $m$ компонентами, для якого $a_i$ дорівнює доходу (в доларах) на один перегляд реклами клієнтом з групи $i$. Зауважимо, що всі елементи  $R, c, v, a$ є невід'ємними числами.

Задайте власні приклади матриці $R$ і векторів $c, a$. Можете використати штучний інтелект для генерації, або ж вигадати щось своє. Намагайтеся зробити дані реалістичними.

```python
#write your code here
import numpy as np

# --- Initialization ---
# Let m = 4 groups of clients and n = 3 advertising channels
m, n = 4, 3

# R (m x n): Reach matrix (views per dollar)
# Rows: Client groups, Columns: Channels
R = np.array([
    [30.0, 10.0, 5.0],   # Group 1: Seniors
    [0.005, 50.0, 20.0],  # Group 2: Students
    [15.0, 5.0, 40.0],   # Group 3: Parents
    [10.0, 15.0, 10.0]    # Group 4: Professionals
])

# c (n): Advertising costs per channel
costs_c = np.array([1000, 500, 1200])

# a (m): Income (in dollars) per one view for each client group
income_per_view_a = np.array([0.05, 0.01, 0.10, 0.08])

# --- Task 1: Total Spending ---
# Formula: Sum of components of vector c, or 1^T * c
ones_n = np.ones(n)
total_spending = ones_n.T @ costs_c
print(f"1. Total spending on advertising: ${total_spending:.2f}")

# --- Task 2: Reach Vector (v) ---
# Formula: v = R * c
# v characterizes total views per client group across all channels
reach_v = R @ costs_c
print(f"\n2. Total reach vector (v):\n{reach_v}")

# --- Task 3: Total Income ---
# Formula: Total Income = a^T * v = a^T * (R * c)
total_income = income_per_view_a.T @ reach_v
print(f"\n3. Total expected income: ${total_income:.2f}")

# --- Task 4: Best Channel for Each Group ---
# Find the index of the maximum value in each row of matrix R
best_channels = np.argmax(R, axis=1)

print("\n4. Most effective channels per group:")
for group_idx, channel_idx in enumerate(best_channels):
    print(f"   Group {group_idx + 1}: Channel {channel_idx + 1} "
          f"(Reach: {R[group_idx, channel_idx]} views/$)")
```
